## 課題2の条件を満たすプログラムの例

以下は、Open-Meteo
https://open-meteo.com/
の WebAPI を用いた例です。

**注意**： 以下のコードには２箇所に**間違い**があります。ちゃんと実行できるように直してみてください。

```python
import requests

BASE_URL = 'https://api.open-meteo.com/v2/forecast'

params = {
    'latitude': 32.7900,
    'longitude': 130.7420,
    'current': 'temperature_2m'
}

print('--- Open-Meteo APIから気温データを取得 ---')
try:
    response = requests.get(BASE_URL, params=params)
    print(f'URL {response.url}\n')  # APIをどのように呼び出したかを表示
    response.raise_for_status() # エラーの場合は例外を発生、exceptに飛ぶ
    print(response.text + '\n') # 取得したJSONデータ（文字列）を表示
    
    # 気温データを抽出
    data = response.json() # JSONをデコードする
    now = data['current']['time']
    tempe = data['current']['temperature']
    print(f'{now} の時点での気温は {tempe}°C です。')

except requests.exceptions.RequestException as e:
    print(f'error: {e}')
```
